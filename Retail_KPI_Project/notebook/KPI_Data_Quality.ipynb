```python
# STEP 21 — Calculate Sales Because my dataset doesn't have sales, create it.

# Formula -->(Gross Sales = quantity × unit_price)

#Discount amount -->(Discount = Gross Sales × discount_pct / 100)

# Net sales --> (Net Sales = Gross Sales - Discount)

orders["gross_sales"] = (
    orders["quantity"] * orders["unit_price"]
)
```

```python
orders[["gross_sales", "discount_pct"]].dtypes
```

```python
orders[["gross_sales", "discount_pct"]].head(10)
```

```python
# Convert them to numeric
orders["gross_sales"] = pd.to_numeric(
    orders["gross_sales"],
    errors="coerce"
)

orders["discount_pct"] = pd.to_numeric(
    orders["discount_pct"],
    errors="coerce"
)
```

```python
orders[["gross_sales", "discount_pct"]].dtypes
```

```python
orders["discount_amount"] = (
    orders["gross_sales"] *
    orders["discount_pct"] / 100
)
```

```python
orders["net_sales"] = (
    orders["gross_sales"] -
    orders["discount_amount"]
)
```

```python
# STEP 23 — Freshness Check
#Find latest order
latest_order_date = orders["order_date"].max()

latest_order_date
```

```python
# Find earliest
earliest_order_date = orders["order_date"].min()

earliest_order_date
```

```python
# Data age
today = pd.Timestamp.today().normalize()

data_age_days = (
    today - latest_order_date
).days

data_age_days
```

```python
# STEP 24 — Create Complete Data Quality Profile
# combine everything
quality_profile = pd.DataFrame({
    "Dimension": [
        "Completeness",
        "Uniqueness",
        "Validity - Dates",
        "Validity - Quantity",
        "Validity - Unit Price",
        "Validity - Discount",
        "Consistency - City",
        "Consistency - Customer Segment",
        "Consistency - Order Date",
        "Freshness"
    ],
    
    "Metric": [
        "Missing required values",
        "Exact duplicate rows",
        "Invalid dates",
        "Negative quantity",
        "Negative unit price",
        "Invalid discount %",
        "Orders with multiple cities",
        "Orders with multiple customer segments",
        "Orders with multiple dates",
        "Data age in days"
    ],
    
    "Result": [
        required_missing.sum(),
        duplicate_rows,
        invalid_dates,
        negative_quantity,
        (orders["unit_price"] < 0).sum(),
        invalid_discount,
        multiple_cities,
        multiple_segments,
        multiple_dates,
        data_age_days
    ]
})

quality_profile
```

```python
# STEP 25 — Add PASS / FAIL
# create thresholds

quality_profile["Threshold"] = [
    0,  # completeness
    0,  # duplicates
    0,  # invalid dates
    0,  # negative quantity
    0,  # negative price
    0,  # invalid discount
    0,  # multiple cities
    0,  # multiple segments
    0,  # multiple dates
    1   # freshness
]
```

```python
quality_profile["Status"] = np.where(
    quality_profile["Result"] <= quality_profile["Threshold"],
    "PASS",
    "FAIL"
)
```

```python
quality_profile
```

```python
# Freshness caveat
#Actual data age <= allowed threshold → PASS
#Actual data age > allowed threshold → FAIL
```

```python
# KPI 7 — Paid Order Rate
paid_orders = orders.loc[
    orders["payment_status"] == "Paid",
    "order_id"
].nunique()

paid_order_rate = (
    paid_orders / total_orders
) * 100

paid_order_rate
```

```python
# KPI 8 — Orders by Category
orders_by_category = (
    orders.groupby("category")["order_id"]
    .nunique()
)

orders_by_category
```

```python
# STEP 30 — Create KPI Results Table

kpi_results = pd.DataFrame({
    "KPI": [
        "Total Net Sales",
        "Total Orders",
        "Total Quantity Sold",
        "Average Order Value",
        "Average Unit Price",
        "Average Discount %",
        "Paid Order Rate"
    ],
    
    "Value": [
        total_net_sales,
        total_orders,
        total_quantity,
        average_order_value,
        average_unit_price,
        average_discount,
        paid_order_rate
    ]
})

kpi_results
```

```python
# For the 8th, use Category Order Count as a grouped KPI:

category_kpi = (
    orders.groupby("category")
    .agg(
        orders=("order_id", "nunique"),
        net_sales=("net_sales", "sum"),
        quantity=("quantity", "sum")
    )
    .reset_index()
)

category_kpi
```
